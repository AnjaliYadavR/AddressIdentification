```python
def get_root_directory():
    """
    Get dataset root directory dynamically.

    Priority:
    1. --root command line argument
    2. DATASET_ROOT environment variable
    3. Current working directory
    """

    parser = argparse.ArgumentParser(
        description="Lease Address Identification Pipeline"
    )

    parser.add_argument(
        "--root",
        type=str,
        required=False,
        help="Root directory containing Excel and PDF files"
    )

    root = Path.cwd()

    root = root.resolve()

    if not root.exists():
        raise FileNotFoundError(
            f"Root directory does not exist: {root}"
        )

    return root
```
